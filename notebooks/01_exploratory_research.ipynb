# SPY / E-mini futures exploratory research

This notebook downloads the public daily proxy, checks alignment, and runs descriptive, stationarity, and cointegration diagnostics. It is research only; it does not submit orders.

```python
from etf_futures_rv.config import DataConfig, StrategyConfig
from etf_futures_rv.data import download_public_data
from etf_futures_rv.research import descriptive_statistics, fair_value_residual, stationarity_test, cointegration_test

prices = download_public_data(DataConfig())
config = StrategyConfig()
descriptive_statistics(prices, config)
```

```python
residual = fair_value_residual(prices, config)
print(stationarity_test(residual))
print(cointegration_test(prices))
residual.mul(10_000).plot(title='Carry residual (bps)')
```
