# Baseline strategy: chronological out-of-sample and walk-forward analysis

Parameters are fixed before this evaluation. Do not tune them on the test set.

```python
from etf_futures_rv.config import BacktestConfig, DataConfig, StrategyConfig
from etf_futures_rv.data import download_public_data
from etf_futures_rv.strategy import generate_signal
from etf_futures_rv.backtest import run_backtest
from etf_futures_rv.metrics import performance_metrics
from etf_futures_rv.validation import chronological_split, walk_forward_returns

prices = download_public_data(DataConfig())
strategy_config, backtest_config = StrategyConfig(), BacktestConfig()
train, test = chronological_split(prices)
signal = generate_signal(prices, strategy_config)
oos_result = run_backtest(test, signal.loc[test.index], backtest_config)
performance_metrics(oos_result)
```

```python
wf_returns = walk_forward_returns(prices, strategy_config, backtest_config)
wf_returns.add(1).cumprod().plot(title='Walk-forward OOS equity index')
print('Walk-forward observations:', len(wf_returns))
```
